# Testing the parsing file by creating a dataframe from it

```python
import pandas as pd

from pathlib import Path
from anomaly_detection.parser.data_parsers import parse_system, parse_security, parse_application
```

```python
# Path to the data
project_folder = Path.cwd().parent

output_path = project_folder / "data/parsing"

security_evtx_path = project_folder / "data/raw/93_securitelog.evtx"
application_evtx_path = project_folder / "data/raw/93_applog.evtx"
system_evtx_path = project_folder / "data/raw/93_syslog.evtx"

application_evtx_path
```

# System

```python
system_training_records = parse_system(system_evtx_path)

system_training_records[0]
```

```python
flat_system_training_record = [{**record['timestamp'], **record['profile'], **record['data']} for record in system_training_records]

flat_system_training_record[0]
```

```python
system_training_df = pd.DataFrame(flat_system_training_record)

system_training_df.head(10)
```

```python
system_encoding_types = [
    "None",
    "None",
    "None",
    "None",
    "None",
    "Convert to int",
    "None",
    "None",
    "None",
    "None",
    "None",
    "None",
    "None",
    "Presence encoding",
    "None",
    "None",
    "None",
    "None",
    "Mixed value encoding",
    "One-hot encoding",
    "Entity encoding",
    "Frequency encoding",
    "One-hot encoding",
    "Mixed value encoding",
    "One-hot encoding",
    "None",
]
```

```python
system_training_summary = pd.DataFrame({
    "column": system_training_df.columns,
    "dtype": system_training_df.dtypes.astype(str),
    "nunique": system_training_df.nunique(dropna=True),
    "nulls": system_training_df.isna().sum(),
    "encoding": system_encoding_types
}).reset_index(drop=True)

system_training_summary.to_csv(output_path / "parsed_system_profile.csv")

system_training_summary
```
